```python
numeric_features = [
    "Port_Congestion",
    "geopolitical_risk",
    "weather_severity",
    "Supplier_Reliability",
    "Supplier_Capacity_Utilization",
    "Inventory_Level",
    "Demand_Volatility",
    "Lead_Time_Days",
    "Shipping_Cost_Index"
]

corr = df[numeric_features].corr()

corr
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Numerical Features")
plt.show()
```

```python
risk_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["Risk_Level_Numeric"] = df["Risk_Level"].map(risk_mapping)

risk_corr = df[numeric_features + ["Risk_Level_Numeric"]].corr()

risk_corr["Risk_Level_Numeric"].sort_values(ascending=False)
```
